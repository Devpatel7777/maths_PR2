```python
total = len(df)
event1 = df[df["final_exam_pass"] == "Pass"]
event2 = df[df["attendance"] > 80]
event3 = df[(df["group_discussion"] == "Yes") & 
            (df["final_exam_pass"] == "Pass")]

print("Probabilities")
print("P(Student Pass):", len(event1)/total)
print("P(attendance > 80):", len(event2)/total)
print("P(Group Discussion and Pass):", len(event3)/total)
```

```python
# Empirical Probability
empirical_pass = len(df[df["final_exam_pass"] == "Pass"]) / total

# Theoretical Probability Example
theoretical_prob = 0.5
print("Empirical vs Theoretical")
print("Empirical P(pass):", empirical_pass)
print("Theoretical Example (coin toss):", theoretical_prob)
```

```python
#RANDOM VARIABLE

# X = number of students passing out of 3 students

p = empirical_pass
from math import comb
print("Probability Distribution")

for x in range(4):
    prob = comb(3, x) * (p**x) * ((1-p)**(3-x))
    print(f"P(X={x}) = {prob:.4f}")

# Mean and Variance
mean = 3 * p
variance = 3 * p * (1 - p)

print("\nMean:", mean)
print("Variance:", variance)
```

```python
#VENN CONDITIONS
A = df["study_hours"] > 10
B = df["attendance"] > 80

print("Venn Values")
print("Only A:", sum(A & ~B))
print("Only B:", sum(B & ~A))
print("Both Conditions:", sum(A & B))
```

```python
#CONTINGENCY TABLE
ct = pd.crosstab(df["group_discussion"], df["final_exam_pass"])
print("Contingency Table")
print(ct)

#Joint Probability
joint = ct.loc["Yes", "Pass"] / total

#Marginal Probability
marginal = ct["Pass"].sum() / total

#Conditional Probability
conditional = ct.loc["Yes", "Pass"] / ct.loc["Yes"].sum()
print("\nJoint Prob(Discussion AND Pass):", joint)
print("Marginal Prob(Pass):", marginal)
print("Conditional Prob(Pass | Discussion):", conditional)
```

```python
#RELATIONSHIP

p_discussion = ct.loc["Yes"].sum() / total
if abs(joint - (p_discussion * marginal)) < 0.05:
    print("Events are Independent")
else:
    print("Events are Dependent")
```

```python
#BAYES THEOREM
P_H_given_pass = 0.7
P_H_given_fail = 0.4

P_pass = empirical_pass
P_fail = 1 - P_pass

# Total Probability
P_H = (P_H_given_pass * P_pass) + (P_H_given_fail * P_fail)

# Bayes Formula
P_pass_given_H = (P_H_given_pass * P_pass) / P_H

print("Bayes Theorem")
print("P(Pass | High Attendance):", P_pass_given_H)
```
